```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
months = np.array([1,2,3,4,5,6])
sales = np.array([200,350,300,500,450,600])
plt.plot(months,sales)
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```
