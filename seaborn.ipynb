```python
import seaborn as sns
import matplotlib.pyplot as plt

import pandas as pd 
data = {
    "Students": ["A","B","C"],
    "Marks": [80,90,70]

}
df = pd.DataFrame(data)
print(df)
sns.barplot(x="Students",y="Marks",data = df)
plt.show()
```

```python
data = [1,2,2,3,3,3,4,4,5]

sns.histplot(data)
plt.show()
```

```python
sns.scatterplot(x=[1,2,3],y=[4,5,6])
plt.show()
```

```python
sns.boxplot(data=[10,20,30,100])
plt.show()
```

```python
data={
    "Math":[90,80,70],
    "Science":[85,75,95],
    "English":[88,76,89]
    
}
df = pd.DataFrame(data)

sns.heatmap(df.corr(),annot=True)
plt.show()
```
