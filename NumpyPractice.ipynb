```python
arr = np.array([[1,10],[3,30],[2,20],[4,40]])
print(arr[arr[:,0]>2])
```

```python
a = np.array([1,2,3])
b = a
b[0] = 100
print(b)
print(a)

```

```python
a = np.array([1,2,3])
b = np.copy(a)
b[0] = 100
print(b)
print(a)
```
